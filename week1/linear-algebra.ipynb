### Vector-vector multipllication (dot product)

row dp column = you need to transpose the row so that you can multiply them

```python
## creation of a function based on this formula 

def vector_vector_multiplication(u ,v): # u and v are the arrays
    assert u.shape[0] == v.shape[0] # making sure they have equal array size

    n = u.shape[0]
    print(f'this is the shape of the array {n}')

    result = 0.0 # will be the answer of the dot product

    for i in range(n):
        result = result + ( u[i] * v[i] )
        print(f'the product is {result}')

    return  result
```

```python
vector_vector_multiplication(np.array([1,0,3]), np.array([4,4,4]))
```

### Matrix-vector multiplication

```python
def matrix_vector_multiplication(U,v): # use capital U to represent matrix
    assert U.shape[1] == v.shape[0]     # the number of columns of the matrix should equal the number of rows of the vector

    num_rows = U.shape[0]
    print(f'the rows of the matrix is {num_rows}')

    result = np.zeros(num_rows) # result is vector in the shape of num_rows    

    for i in range(num_rows):
        result[i] = vector_vector_multiplication(U[i], v)

    return result


     
```

```python
u = [ [2,3,4],
     [5,3,1],
     [7,1,2],
]

u = np.array(u)
v = np.array([4,7,2])
```

```python
matrix_vector_multiplication(u,v)
```

```python
## using a built-in function instead

u.dot(v)
```
